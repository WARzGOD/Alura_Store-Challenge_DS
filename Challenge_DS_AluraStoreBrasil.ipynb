### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento



```python
faturamento_loja = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

print(f"Faturamento Loja 1: R$ {faturamento_loja:,.2f}")
print(f"Faturamento Loja 2: R$ {faturamento_loja2:,.2f}")
print(f"Faturamento Loja 3: R$ {faturamento_loja3:,.2f}")
print(f"Faturamento Loja 4: R$ {faturamento_loja4:,.2f}")
```

```python
import matplotlib.pyplot as plt

faturamentos = [faturamento_loja, faturamento_loja2, faturamento_loja3, faturamento_loja4]
lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

cores = ['#4CAF50', '#2196F3', '#FFC107', '#FF5722']

plt.figure(figsize=(8, 8))
plt.pie(
    faturamentos,
    labels=lojas,
    autopct='%1.1f%%',
    startangle=90,
    colors=cores,
    shadow=True,
    explode=(0.03, 0.03, 0.03, 0.03)
)


plt.title('Distribuição do Faturamento Total por Loja', fontsize=14, fontweight='bold')
plt.axis('equal')
plt.show()
```

# 2. Vendas por Categoria

```python
vendas_cat1 = loja['Categoria do Produto'].value_counts()
vendas_cat2 = loja2['Categoria do Produto'].value_counts()
vendas_cat3 = loja3['Categoria do Produto'].value_counts()
vendas_cat4 = loja4['Categoria do Produto'].value_counts()

print("=== Vendas Loja 1 ===")
display(vendas_cat1)
print("=== Vendas Loja 2 ===")
display(vendas_cat2)
print("=== Vendas Loja 3 ===")
display(vendas_cat3)
print("=== Vendas Loja 4 ===")
display(vendas_cat4)
```

```python
import matplotlib.pyplot as plt


lojas = [vendas_cat1, vendas_cat2, vendas_cat3, vendas_cat4]
titulos = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

plt.figure(figsize=(14, 10))

for i in range(4):
    plt.subplot(2, 2, i+1)
    lojas[i].plot(kind='bar', color='#2196F3')
    plt.title(f'Vendas por Categoria - {titulos[i]}', fontsize=12, fontweight='bold')
    plt.xlabel('Categoria')
    plt.ylabel('Quantidade de Produtos Vendidos')
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

# 3. Média de Avaliação das Lojas


```python
media_avaliacao_loja1 = loja['Avaliação da compra'].mean()
media_avaliacao_loja2 = loja2['Avaliação da compra'].mean()
media_avaliacao_loja3 = loja3['Avaliação da compra'].mean()
media_avaliacao_loja4 = loja4['Avaliação da compra'].mean()

print(f"Média de avaliação - Loja 1: {media_avaliacao_loja1:.2f}")
print(f"Média de avaliação - Loja 2: {media_avaliacao_loja2:.2f}")
print(f"Média de avaliação - Loja 3: {media_avaliacao_loja3:.2f}")
print(f"Média de avaliação - Loja 4: {media_avaliacao_loja4:.2f}")
```

```python
import matplotlib.pyplot as plt

media_avaliacao_loja1 = loja['Avaliação da compra'].mean()
media_avaliacao_loja2 = loja2['Avaliação da compra'].mean()
media_avaliacao_loja3 = loja3['Avaliação da compra'].mean()
media_avaliacao_loja4 = loja4['Avaliação da compra'].mean()

lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
medias = [media_avaliacao_loja1, media_avaliacao_loja2, media_avaliacao_loja3, media_avaliacao_loja4]

plt.figure(figsize=(8,5))
plt.barh(lojas, medias, color=['#4CAF50', '#2196F3', '#FFC107', '#F44336'])

for i, v in enumerate(medias):
    plt.text(v + 0.05, i, f'{v:.2f}', va='center', fontsize=11)

plt.title('Média de Avaliação por Loja - Alura Store', fontsize=14, fontweight='bold')
plt.xlabel('Média das Avaliações (0 a 5)', fontsize=12)
plt.ylabel('Lojas', fontsize=12)
plt.xlim(0, 5)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()

```

# 4. Produtos Mais e Menos Vendidos

```python
import matplotlib.pyplot as plt

# --- LOJA 1 ---
plt.figure(figsize=(8,5))
mais_vendidos_1.head(5).sort_values().plot(kind='barh', color='blue')
plt.title('Top 5 Produtos Mais Vendidos - Loja 1')
plt.xlabel('Quantidade Vendida')
plt.ylabel('Produto')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.show()

# --- LOJA 2 ---
plt.figure(figsize=(8,5))
mais_vendidos_2.head(5).sort_values().plot(kind='barh', color='green')
plt.title('Top 5 Produtos Mais Vendidos - Loja 2')
plt.xlabel('Quantidade Vendida')
plt.ylabel('Produto')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.show()


# --- LOJA 3 ---
plt.figure(figsize=(8,5))
mais_vendidos_3.head(5).sort_values().plot(kind='barh', color='orange')
plt.title('Top 5 Produtos Mais Vendidos - Loja 3')
plt.xlabel('Quantidade Vendida')
plt.ylabel('Produto')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.show()


# --- LOJA 4 ---
plt.figure(figsize=(8,5))
mais_vendidos_4.head(5).sort_values().plot(kind='barh', color='purple')
plt.title('Top 5 Produtos Mais Vendidos - Loja 4')
plt.xlabel('Quantidade Vendida')
plt.ylabel('Produto')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.show()

```

# 5. Frete Médio por Loja

```python
frete_medio_loja1 = loja['Frete'].mean()
frete_medio_loja2 = loja2['Frete'].mean()
frete_medio_loja3 = loja3['Frete'].mean()
frete_medio_loja4 = loja4['Frete'].mean()

print(f"Frete médio - Loja 1: R$ {frete_medio_loja1:.2f}")
print(f"Frete médio - Loja 2: R$ {frete_medio_loja2:.2f}")
print(f"Frete médio - Loja 3: R$ {frete_medio_loja3:.2f}")
print(f"Frete médio - Loja 4: R$ {frete_medio_loja4:.2f}")
```

```python
import matplotlib.pyplot as plt

fretes_medios = [frete_medio_loja1, frete_medio_loja2, frete_medio_loja3, frete_medio_loja4]
lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']

plt.figure(figsize=(8,5))
plt.plot(lojas, fretes_medios,
         marker='o',
         linestyle='-',
         color='#2196F3',
         linewidth=2,
         markersize=8)

for i, valor in enumerate(fretes_medios):
    plt.text(lojas[i], valor + 0.1, f"R$ {valor:.2f}", ha='center', fontsize=10, fontweight='bold')

plt.title('Frete Médio por Loja - Alura Store', fontsize=14, fontweight='bold')
plt.xlabel('Lojas', fontsize=12)
plt.ylabel('Frete Médio (R$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

##Relatório de Análise de Lojas – Alura Store
###Introdução

O objetivo deste relatório é fornecer ao Senhor João uma base analítica para decidir qual loja da rede Alura Store deve ser vendida. Para isso, foram analisados os seguintes fatores para as quatro lojas da rede:

1. Faturamento total.

2. Categorias de produtos mais e menos vendidas.

3. Média de avaliação dos clientes.

4. Produtos mais e menos vendidos.

5. Frete médio por loja.

As análises foram complementadas com gráficos de barras, pizza, linha e boxplot, a fim de tornar os resultados mais claros e visualmente compreensíveis.

###Desenvolvimento
1. **Faturamento Total**

O faturamento total de cada loja foi calculado somando-se os preços de todos os produtos vendidos.

Observou-se que a Loja 4 apresentou o menor faturamento, enquanto a Loja 1 apresentou o maior faturamento.

O gráfico de pizza mostrou que a Loja 4 contribui com uma proporção menor do faturamento total da rede, indicando menor desempenho financeiro.

2. **Categorias de Produtos Mais e Menos Vendidas**

As vendas foram analisadas por categoria:

Todas as lojas apresentam categorias mais vendidas semelhantes, como eletrônicos e acessórios.

Contudo, a Loja 3 apresenta vendas concentradas em poucas categorias, com menor diversificação, enquanto a Loja 2 tem vendas mais equilibradas entre categorias.

As barras horizontais e os gráficos de pizza destacam que a Loja 3 possui categorias pouco representativas, o que pode limitar o público-alvo.

3. **Média de Avaliação dos Clientes**

A média das avaliações de clientes foi calculada para cada loja:

Loja 1 apresentou a menor média de avaliação, indicando menor satisfação dos clientes. Já a loja 3 apresentou a maior média, indicando boa satisfação dos clientes.

As análises com boxplot mostraram que a Loja 1 também possui mais avaliações negativas ou muito abaixo da média, reforçando a percepção de desempenho inferior.

4. **Produtos Mais e Menos Vendidos**

A análise dos produtos mais vendidos demonstrou que a Loja 3 depende de poucos produtos, com uma concentração de vendas maior em itens específicos, evidenciada pelo gráfico de Pareto.

Além disso, os produtos menos vendidos na Loja 3 representam uma parcela significativa do estoque não movimentado, diferentemente das outras lojas que possuem um mix mais equilibrado de vendas.

5. **Frete Médio**

O frete médio da Loja 1 é mais alto em relação às outras lojas, o que pode reduzir ainda mais a atratividade das vendas.

Ademais, o gráfico mostra que a loja 4 é a que possui o menor frete,o que pode atrair mais os clientes para produtos específicos dessaa loja.

###**Conclusão e Recomendação**

Com base nas análises realizadas:

Apesar do maior faturamento, a **Loja 1** apresenta vários indicadores de ineficiência e insatisfação dos clientes:

- É a loja com pior média de avaliação, conforme o gráfico de barras horizontais.

- Apresenta maior custo médio de frete, o que diminui sua competitividade e pode explicar parte da insatisfação dos consumidores.

- O alto faturamento pode estar relacionado a preços maiores, e não necessariamente a um bom desempenho operacional.

####Do ponto de vista qualitativo, a Loja 1:

- Tem maior incidência de avaliações negativas, sugerindo problemas com atendimento, entrega ou qualidade de produto.

- Pode estar sustentando o faturamento alto com margens mais baixas ou custos altos de logística (frete).

####Comparativamente:

- A Loja 3 se destaca positivamente em satisfação dos clientes (melhor média de avaliação) e possui frete mais equilibrado, ainda que tenha menor diversificação de produtos.

- A Loja 2 é a mais equilibrada, com bom desempenho em todas as métricas.

- A Loja 4 tem o menor faturamento, mas bom custo de frete — pode crescer com estratégias de marketing e diversificação.

Conclusão: A **loja 1** deve ser vendida!!
